## 2. POS Tagging and Token Analysis

Process a paragraph of text and print detailed information about each token.

For each token, you need to access:

| Property | What It Gives You | Example |
|----------|-------------------|--------|
| `token.text` | The word itself | "cat" |
| `token.lemma_` | The base form of the word | "sat" → "sit" |
| `token.pos_` | Part of speech | "NOUN", "VERB", "ADJ" |
| `token.dep_` | Dependency label | "nsubj", "ROOT", "dobj" |
| `token.head` | The word this token depends on | "sat" |

### POS Tags Reference

**Universal POS Tags (`token.pos_`)**

| Tag | Meaning | Example |
|-----|---------|---------|
| ADJ | Adjective | big, old, green |
| ADP | Adposition (preposition) | in, on, at, to |
| ADV | Adverb | very, quickly, never |
| AUX | Auxiliary verb | is, has, will, should |
| CCONJ | Coordinating conjunction | and, but, or |
| DET | Determiner | the, a, this, every |
| INTJ | Interjection | oh, wow, oops |
| NOUN | Noun | cat, tree, software |
| NUM | Numeral | one, 2024, third |
| PART | Particle | not, 's, up (in "give up") |
| PRON | Pronoun | I, you, he, she, it |
| PROPN | Proper noun | Google, London, Mary |
| PUNCT | Punctuation | . , ! ? |
| SCONJ | Subordinating conjunction | if, because, that |
| SYM | Symbol | $, %, +, = |
| VERB | Verb | run, eat, published |
| X | Other | foreign words, typos |



```python
# A short paragraph to analyze
text = (
    "The curious cat sat on the mat. "
    "Then it jumped onto the table and knocked over a glass of water. "
    "Its owner laughed and took a picture."
)

# TODO 1: Process the text through spaCy to create a doc object
# Hint: doc = nlp(____)
doc = nlp(text)

# TODO 2: Complete the print statement for each token
# Fill in the four blanks with the correct token properties
print('=== TOKEN DETAILS ===')
print(f'{"WORD":<12} {"LEMMA":<12} {"POS":<8} {"DEP":<10} {"HEAD"}')
print('-' * 55)

for token in doc:
    print(
        f"{token.text:<12} "
        f"{token.lemma_:<12} "
        f"{token.pos_:<8} "
        f"{token.dep_:<10} "
        f"{token.head.text}"
    )
```

## 3. Extract Subject-Verb-Object (SVO) Triples

For each sentence, find:

- The **main verb** -- the token with `dep_` equal to `"ROOT"`
- The **subject** -- a child of the verb with `dep_` equal to `"nsubj"` or `"nsubjpass"`
- The **object** -- a child of the verb with `dep_` equal to `"dobj"`, `"pobj"`, `"attr"`, or `"oprd"`

Every token has a `.children` property that gives you the tokens that depend on it.

For example, in "The cat sat on the mat":
- "sat" is the ROOT (main verb)
- "cat" is the nsubj (subject) of "sat"
- "mat" connects back through "on" as a prepositional object

### Dependency Labels Reference

**Common Dependency Labels (`token.dep_`)**

| Label | Meaning | Example |
|-------|---------|---------|
| ROOT | Main verb of the sentence | "The cat **sat**" |
| nsubj | Nominal subject | "The **cat** sat" |
| nsubjpass | Passive subject | "The **cat** was chased" |
| dobj | Direct object | "She ate **pizza**" |
| iobj | Indirect object | "She gave **him** a book" |
| pobj | Object of preposition | "sat on the **mat**" |
| attr | Attribute | "He is a **teacher**" |
| det | Determiner | "**The** cat" |
| amod | Adjective modifier | "the **big** cat" |
| advmod | Adverb modifier | "ran **quickly**" |
| prep | Preposition | "sat **on** the mat" |
| aux | Auxiliary | "**will** run" |
| neg | Negation | "did **not** go" |
| cc | Coordinating conjunction | "cats **and** dogs" |
| conj | Conjunct | "cats and **dogs**" |
| compound | Compound word | "**ice** cream" |
| punct | Punctuation | **.** |

```python
print('=== SVO TRIPLES ===')
print()

for sent in doc.sents:
    main_verb = None
    subject = None
    obj = None

    # TODO 3: Find the main verb (ROOT) in this sentence
    # Loop through the tokens in 'sent'
    # Look for the token where dep_ == "ROOT"
    for token in sent:
        if token.dep == 'ROOT':
            main_verb = token
            break

    # TODO 4: Search the main verb's children for subject and object
    # subject dep_ is "nsubj" or "nsubjpass"
    # object dep_ is "dobj", "pobj", "attr", or "oprd"
    if main_verb is not None:
        for child in main_verb.children:
            if child.dep_ in ('nsubj', 'nsubjpass'):
                subject = child
            if child.dep_ in ('dobj', 'pobj', 'attr', 'oprd'):
                obj = child

    # Print the result (given)
    print(f'Sentence: "{sent.text.strip()}"')
    s = subject.text if subject else '-'
    v = main_verb.text if main_verb else '-'
    o = obj.text if obj else '-'
    print(f'  Subject: {s}  |  Verb: {v}  |  Object: {o}')
    print()
```
