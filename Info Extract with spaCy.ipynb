## 6. Visualize the Dependency Parse Tree

The dependency parse tree shows how every word in a sentence connects to every other word grammatically.
Arcs point from each word to the word it depends on.

```python
simple_doc = nlp("The curious cat sat on the mat.")

displacy.render(simple_doc, style='dep', jupyter=True)
```
