```python
import pandas as pd
from transformers import AutoTokenizer

TERMS = [
    # term, category   (Table 1 of the PubMedBERT paper)
    ("diabetes",          "disease"),
    ("leukemia",          "disease"),
    ("lithium",           "drug"),
    ("insulin",           "drug"),
    ("DNA",               "gene"),
    ("promoter",          "gene"),
    ("hypertension",      "disease"),
    ("nephropathy",       "disease"),
    ("lymphoma",          "disease"),
    ("lidocaine",         "drug"),
    ("oropharyngeal",     "organ"),
    ("cardiomyocyte",     "cell"),
    ("chloramphenicol",   "drug"),
    ("RecA",              "gene"),
    ("acetyltransferase", "gene"),
    ("clonidine",         "drug"),
    ("naloxone",          "drug"),
]

MODELS = [
    ("BERT",       "bert-base-uncased"),
    ("SciBERT",    "allenai/scibert_scivocab_uncased"),
    ("BioBERT",    "dmis-lab/biobert-v1.1"),
    ("PubMedBERT", "microsoft/BiomedNLP-BiomedBERT-base-uncased-abstract-fulltext"),
]

tokenizers = {}
for label, repo in MODELS:
    tokenizers[label] = AutoTokenizer.from_pretrained(repo)
    print(label, "vocabulary:", tokenizers[label].vocab_size)


def as_paper(pieces):
    """Write a token list the way the paper writes it: whole, or the pieces joined by hyphens."""
    if len(pieces) == 1:
        return "whole"
    return "-".join(pieces)


def highlight(rows, columns):
    """Show the table with the named columns in bold, so it reads without relying on colour."""
    table = pd.DataFrame(rows)
    return table.style.set_properties(subset=columns, **{"font-weight": "bold"})


rows = []
for term, category in TERMS:
    row = {"Biomedical Term": term, "Category": category}
    for label, repo in MODELS:
        row[label] = as_paper(tokenizers[label].tokenize(term))
    rows.append(row)

pd.DataFrame(rows)

# Expected: the paper's comparison, reproduced from the tokenizers themselves.
# These models are uncased, so RecA and DNA arrive lowercased.
# Look hard at the BioBERT column before moving on.
```

```python
# TODO: load gpt2, call it baselineGPT
# hint: AutoTokenizer is already imported above. Part 5 loaded it the same way, one line.
baselineGPT = AutoTokenizer.from_pretrained("gpt2")

try:
    # TODO: how many tokens does baselineGPT know
    # hint: baselineGPT.something, no brackets. Part 5 printed it.
    print("baselineGPT vocabulary:", baselineGPT.vocab_size)

    for row in rows:
        term = row["Biomedical Term"]

        # the term on its own, as the paper tested it
        # TODO: turn term into its tokens
        # hint: baselineGPT.something(term). Part 5 used it on add_numbers.
        pieces = baselineGPT.tokenize(term)

        # the same term as it would arrive mid sentence, with a space in front of it
        # TODO: the same call again, on the term with a space in front of it
        # hint: " " + term
        spaced = baselineGPT.tokenize(" " + term)

        row["GPT-2"] = as_paper(pieces)
        row["GPT-2 with space"] = as_paper(spaced)

    display(highlight(rows, ["GPT-2", "GPT-2 with space"]))
except NameError:
    print("Fill in the blanks above, then run this cell again.")

# Expected: 50257, then the same table with two GPT-2 columns added.
# The two GPT-2 columns will not agree. That is question 2.
```

```python
from datasets import load_dataset

DATASET = "qiaojin/PubMedQA"   # the dataset, as it is named on the Hub
CONFIG = "pqa_labeled"         # the expert labelled part of it


def load_corpus():
    """Return the train split of the labelled part of PubMedQA."""
    # TODO: return the loaded dataset
    # hint: load_dataset(<dataset>, <configuration>, split=<split name>)
    #       the first two are the variables above, and the split you want is called "train"
    return load_dataset(DATASET, CONFIG, split='train')


try:
    raw = load_corpus()

    print("rows  :", len(raw))
    print("fields:", raw.column_names)
except NameError:
    print("Fill in the blank inside load_corpus above, then run this cell again.")

# Expected: 1000 rows, then the column names.
```

```python
# each row's context holds a list of paragraphs; join them into one passage
passages = []
try:
    for row in raw:
        passages.append(" ".join(row["context"]["contexts"]))

    print("passages:", len(passages))
    print()
    print(passages[0][:600])
except NameError:
    print("Load the dataset in the cell above first, then run this cell again.")

# Expected: 1000 passages, then the start of one abstract.
```

```python
def get_training_corpus(batch_size=100):
    """Yield the passages in batches, so the whole corpus is never copied."""
    for start in range(0, len(passages), batch_size):
        yield passages[start:start + batch_size]


for batch in get_training_corpus():
    print("first batch:", len(batch), "passages")
    break

# Expected: first batch: 100 passages
```

```python
# TODO: choose a size between 10000 and 30000
VOCAB_SIZE = 20000


def retrain(vocab_size):
    """Train a new tokenizer from baselineGPT on the PubMedQA passages."""
    # TODO: return the newly trained tokenizer
    # hint: baselineGPT.something(get_training_corpus(), vocab_size). Part 5 ran it.
    return baselineGPT.train_new_from_iterator(get_training_corpus(), vocab_size)


if VOCAB_SIZE == "TODO":
    print("Choose a VOCAB_SIZE above, then run this cell again.")
else:
    try:
        biomedGPT = retrain(VOCAB_SIZE)

        print("asked for  :", VOCAB_SIZE)
        print("baselineGPT:", baselineGPT.vocab_size)
        print("biomedGPT  :", biomedGPT.vocab_size)
    except NameError:
        print("Fill in the blank inside retrain above, then run this cell again.")

# Expected: your number, 50257, then the size reached.
# Reaching less is possible. A few hundred means the generator was reused.
```

```python
def tokens_per_word(text, tokenizer):
    """Return the number of tokens, the number of words, and the ratio."""
    # TODO: how many token ids does text produce
    # hint: len(tokenizer(text).something). Part 5 counted them this way.
    n_tokens = len(tokenizer(text).input_ids)

    n_words = len(text.split())
    return n_tokens, n_words, n_tokens / n_words


ENGLISH_SAMPLE = ("The tokenizer is chosen and frozen before the model is trained, which is why a "
                  "pretrained model can never be given a different one later on.")

import pandas as pd

try:
    ABSTRACT_SAMPLE = passages[-1]

    rows = []
    for name, tok in [("baselineGPT", baselineGPT), ("biomedGPT", biomedGPT)]:
        for label, body in [("abstract", ABSTRACT_SAMPLE), ("english", ENGLISH_SAMPLE)]:
            n_tokens, n_words, ratio = tokens_per_word(body, tok)
            rows.append({
                "tokenizer": name,
                "sample": label,
                "tokens": n_tokens,
                "words": n_words,
                "tokens per word": round(ratio, 2),
            })

    table = pd.DataFrame(rows)
    display(table)
except (NameError, IndexError):
    print("Complete Exercises 2 and 3 first, then fill in the blank above and run this cell again.")

# Expected: four rows. biomedGPT cheaper on the abstract, dearer on the english.
# Winning both means Exercise 3 went wrong.
```

```python
import pandas as pd

try:
    changed = []
    for term, category in TERMS:
        before = baselineGPT.tokenize(term)
        after = biomedGPT.tokenize(term)
        changed.append({
            "term": term,
            "baselineGPT pieces": before,
            "biomedGPT pieces": after,
            "before": len(before),
            "after": len(after),
            "saved": len(before) - len(after),
        })

    display(pd.DataFrame(changed))

    biomedGPT.save_pretrained("pubmedqa-tokenizer")
    print("saved to pubmedqa-tokenizer")
except NameError:
    print("Complete Exercises 1 and 3 first, then run this cell again.")

# Expected: seventeen rows, both splits written out, then the save line.
# Several should improve. Some may get worse, which is a real result worth reporting.
```

```python
import pandas as pd

try:
    final = []
    for term, category in TERMS:
        row = {"Biomedical Term": term, "Category": category}
        for label, repo in MODELS:
            row[label] = as_paper(tokenizers[label].tokenize(term))
        row["GPT-2"] = as_paper(baselineGPT.tokenize(term))
        row["biomedGPT (ours)"] = as_paper(biomedGPT.tokenize(term))
        final.append(row)

    display(highlight(final, ["biomedGPT (ours)"]))
except NameError:
    print("Complete Exercises 1 and 3 first, then run this cell again.")

# Expected: seventeen rows, eight columns, the last one bold.
# biomedGPT sits between GPT-2 and PubMedBERT on most terms. It was trained on a thousand passages.
```
